```python
import numpy as np

def word_lengths(texts):
    return np.array([len(t.split()) for t in texts], dtype=np.int32)

len_head = word_lengths(texts_headline)
len_body = word_lengths(texts_body)

print("Headline: p50/p90/p95 =", np.percentile(len_head, [50, 90, 95]))
print("Text:     p50/p90/p95 =", np.percentile(len_body, [50, 90, 95]))
```
